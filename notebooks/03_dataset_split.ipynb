```python
import os

import pandas as pd
from sklearn.model_selection import train_test_split
```

```python
# Paths to the data directory
DIR_TRAIN = '../data/raw/aptos2019/train_images'
# Read the training set metadata
df = pd.read_csv('../data/raw/aptos2019/train.csv')
```

Use a stratified split with 70/15/15 proportions. A stratified split ensures that the proportion of each class in the original dataset is preserved during the split. To read more about it: https://medium.com/@becaye-balde/why-you-should-use-stratified-split-bddb6dadd34e

```python
train_df, temp_df = train_test_split(
    df,
    test_size = 0.3,
    stratify=df["diagnosis"],
    random_state=42
)
```

```python
val_df, test_df = train_test_split(
    temp_df,
    test_size = 0.5,
    stratify=temp_df["diagnosis"],
    random_state=42
)
```

```python
print("Number of train samples by diagnosis:")
print(train_df["diagnosis"].value_counts())
```

```python
print("Number of val samples by diagnosis:")
print(val_df["diagnosis"].value_counts())
```

```python
print("Number of test samples by diagnosis:")
print(test_df["diagnosis"].value_counts())
```

```python
output_dir = "../data/processed/aptos2019/splits"
os.makedirs(output_dir, exist_ok=True)
```

```python
train_df.to_csv(os.path.join(output_dir, "train_split.csv"), index=False)
val_df.to_csv(os.path.join(output_dir, "val_split.csv"), index=False)
test_df.to_csv(os.path.join(output_dir, "test_split.csv"), index=False)
```
